##  34) ข้อใดต่อไปนี้เป็นการใช้เงื่อนไขหลายเงื่อนไขที่แสดงถึง

##  35) ข้อใดต่อไปนี้เป็นการใช้เงื่อนไขหลายเงื่อนไขที่แสดงถึงเงื่อนไข OR
